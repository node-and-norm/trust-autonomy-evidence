## 2. What problem does the method address?

A signature, job title or approval step cannot alone show that someone had usable control. A later investigator needs to know what the person could access, what options were available, and whether an intervention could reach execution.

The method makes each of those questions visible. It separates three things:

1. **Source evidence:** what the retained record says, including its limits.
2. **Assessor interpretation:** why a passage is assigned a condition-level judgment.
3. **Calculation:** how those judgments produce the declared case finding.

Checking the calculation does not settle the interpretation. Preserving a file hash does not prove that the source is true.

[Question and contribution](https://github.com/node-and-norm/trust-autonomy-evidence/blob/a8bd6a13cf8f24a1abe15885df7195ca20465a25/docs/research-summary.md) · [November 1 source manuscript, sections 1 and 3](https://github.com/node-and-norm/trust-autonomy-evidence/blob/a8bd6a13cf8f24a1abe15885df7195ca20465a25/paper/revisions/2026-11-01/main.tex)

## 3. Follow the six conditions

**Define the decision → assemble the record → assess each condition → retain uncertainty → apply the rule → expose reasons and limits.**

| Condition | Question for the reader |
|:---|:---|
| Access | Was relevant information received before commitment or irreversible action? |
| Comprehension capacity | Could the person use the information, limitations and alternatives under the operating conditions? |
| Authority | Could the person approve, reject, change, stop or escalate? |
| Feasibility | Could that authority be exercised with the available time, staffing and access? |
| Exercise | Is there evidence of active judgment or intervention? |
| Execution propagation | Did an intervention reach the action or obligation? |

These questions summarize, rather than replace, the protocol. A record of access is not a direct observation of private cognition. A changed execution record is not automatically evidence of a beneficial outcome or a counterfactual causal effect.

Correction, repair and reform are three separate post-event stages. Later improvement cannot turn failed or unresolved event-level control into a pass.

[Exact definitions, evidence requirements and rule](https://github.com/node-and-norm/trust-autonomy-evidence/blob/a8bd6a13cf8f24a1abe15885df7195ca20465a25/protocols/practical-human-control-test.md)

### How to read a finding

| State | Meaning in the protocol |
|:---|:---|
| Supported | Direct, contemporaneous evidence satisfies the condition |
| Partially supported | Some required elements are present, but a material gap remains |
| Unsupported | Evidence contradicts the condition or shows it was absent |
| Indeterminate | The packet is insufficient to decide |
| Outside scope | A justified boundary exclusion; it cannot remove a required event condition |

**Fail:** any justified Unsupported necessary condition.  
**Pass:** all six necessary conditions Supported.  
**Unresolved:** no necessary condition Unsupported, but at least one Partial or Indeterminate.

Missing evidence is not automatically evidence of failure. The categorical result is not an overall numerical control score.

[Assessment contract](https://github.com/node-and-norm/trust-autonomy-evidence/blob/a8bd6a13cf8f24a1abe15885df7195ca20465a25/protocols/practical-human-control-test.md)
