## 1 The question and the rule

The bounded case concerns the Patriot classification and engagement of RAF Tornado ZG710. The present example examines **comprehension capacity**, not blame or whether the final outcome was beneficial.

The released criterion asks whether the authority could interpret the output, its limits, uncertainty and alternatives. Supported requires direct contemporaneous evidence. Unsupported requires contradictory evidence or evidence of absence. Indeterminate means the packet is insufficient to decide.

An adverse training finding can matter without establishing the entire inference required for Unsupported. Rejecting Unsupported also does not establish Supported or Partially supported.

[Read the pinned assessment contract](https://github.com/node-and-norm/trust-autonomy-evidence/blob/564406d16f74f66a59ee581f176c9b55f3be5f14/protocols/practical-human-control-test.md).

**Who is being assessed?** The control assessment concerns the Patriot battery crew. The RAF Tornado aircrew were the people killed. These are distinct groups.

## 3 The preserved original judgment

**Historical classification: Unsupported.**

**Exact assessor rationale, not a quotation from the inquiry:**

> Training to trust the system and limited explanatory context prevented a sound interpretation of the false classification.

That wording goes beyond reporting a training deficiency: it says the deficiency prevented sound interpretation in the bounded event. The later review disputes whether that event-level inference is sufficiently established.

## 5 Keep the strongest alternative visible

A reviewer could argue that the documented preparation and information deficiencies defeated a necessary component of usable limitations understanding, making Unsupported appropriate. That argument does not need to prove total incapacity, and evidence of basic operating skill would not automatically defeat it.

The approved reassessment's objection is narrower: the connection from those deficiencies to the Patriot battery crew's necessary capacity remains insufficiently established in the inspected record.

The distinction is not settled by counting citations, by the harmful outcome, or by the successful execution of a launch. It turns on the bounded proposition, admissible evidence and the inference the assessor can defend.

This example preserves the competing interpretations; it does not record a new adjudication. ## 8 A dated metadata review note

**Finding:** the frozen T2 manifest records publication year **2006**, whereas the official PDF cover says **May 2004**. Two additional official records clarify the release chronology:

- [Hansard, 21 May 2004](https://hansard.parliament.uk/Lords/2004-05-21/debates/11dffe42-c426-47af-a5f8-cb7a6d41d916/WrittenStatements), under “Military Aircraft Accident Summary: Tornado Gr4a Zg710,” records that copies of the summary were placed in the House library that day.
- A [Ministry of Defence memorandum dated 27 July 2006](https://publications.parliament.uk/pa/cm200607/cmselect/cmpubacc/486/6031308.htm), table row 1, records internet release in **May 2004**. The memorandum's own 2006 date is not the summary's release date.
- The [current GOV.UK landing page](https://www.gov.uk/government/publications/military-aircraft-accident-summary-aircraft-accident-to-raf-tornado-gr-mk4a-zg710) displays **23 March 2003** as Published. That metadata conflicts with the summary's cover date and the later release records; it should not be used by itself as the summary's publication date.

**Recommended future correction:** retain the historical value in the frozen manifest; add a separate dated annotation identifying May 2004 as the supported document/release month, with the parliamentary record separately dated 21 May. The origin of the recorded 2006 value remains unknown. Do not claim an exact first internet-release day.
