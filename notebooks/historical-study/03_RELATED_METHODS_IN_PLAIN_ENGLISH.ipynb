## 1. Start with the human question

Imagine an automated system is about to take an important action. A person is formally allowed to intervene. Later, an investigator asks:

> Did the person have a practical opportunity to intervene, and what surviving evidence supports that judgment?

Permission alone does not answer that question. Nor does a bad outcome prove that the person lacked control or made a bad decision.

**TAE** is the shorthand used here for the `trust-autonomy-evidence` study.

**TAE's task** is to make its assessment traceable: identify the specific decision and time period being assessed, link judgments to records, apply its stated rules and retain uncertainty. In the current reassessment, all three historical cases remain unresolved under those recorded judgments. That is not proof that control existed or was absent.

[November 1 source manuscript, research question and scope (§§1–2)](https://github.com/node-and-norm/trust-autonomy-evidence/blob/main/paper/revisions/2026-11-01/main.tex)

## 2. Four approaches, four useful questions

These are brief reading aids, **not a ranking or a claim that the methods are interchangeable**.

### TAE: “What can this record support about practical human control?”
The November manuscript applies a frozen assessment procedure to public records for specifically defined cases. It preserves the path from evidence through judgments to a result, including uncertainty and corrections. The reassessment was retrospective, not a new independent test.

**Useful here:** seeing where an assessment depends on an interpretation or missing record.

**Not established:** that different assessors will agree, that its conclusions are more accurate than alternatives, or that using it improves decisions.

[TAE manuscript, §§1.2–2.5 and 5](https://github.com/node-and-norm/trust-autonomy-evidence/blob/main/paper/revisions/2026-11-01/main.tex)

### McCarthy's CAST analysis: “How did the wider system contribute?”
Scott McCarthy's 2013 MIT thesis examines the 2003 Tornado ZG710 friendly-fire event discussed in TAE using CAST, a systems-safety analysis method. It examines the control structure and constraints, rather than reducing the event to one person's mistake.

**Useful here:** a wider account of training, communications and relationships between parts of the system.

**Important limit:** this is an analytical interpretation of records, not independent observation of the event or proof of what an individual thought.

[Thesis record](https://dspace.mit.edu/entities/publication/bf73a6b1-f38f-4620-a4d1-a49855338bed) · [MIT-hosted thesis, Chapters 4–6](https://www.mit.stamp-institute.com/uploads/2013_A_System_Theoretic_Safety_Analysis_of_Fr.pdf)

### CEDAR: “How can an assessor organize evidence for an AI-management audit?”
CEDAR organizes audit obligations, evidence judgments and suggested improvement priorities. It separates the assessor's coding from the deterministic calculations applied to that coding. Insufficient public evidence can remain **Not assessed**.

**Useful here:** it already addresses traceability, missing evidence and repeatable processing. We should not present those ideas alone as TAE's novelty.

**Important limit:** rerunning the same coded inputs is not the same as independent assessors reaching the same judgment. Its retrospective public-record case analysis is not independent validation of the method.

[CEDAR, §§4.1, 5.3, 6 and Appendix A](https://arxiv.org/html/2606.21276v1)

### Niazi and colleagues: “How can rights, risks and rules stay connected to control evidence?”
Their framework uses versioned assurance cases: structured arguments that connect claims, assumptions and evidence. It discusses human override traces and named control points, with retrospective analysis of public materials.

**Useful here:** it connects governance claims with operational evidence and already overlaps with evidence-based human-control reasoning.

**Important limit:** a proposed control mechanism is not proof that it has been deployed or validated. This review has not independently checked its underlying coding artifacts or assessor reliability.

[Rights-Based AI in Cyber–Physical Systems, methodology and assurance-case sections](https://doi.org/10.3390/automation7030096)

**In everyday terms:** CAST looks widely at the system; CEDAR organizes audit evidence; Niazi links governance arguments to control evidence; TAE offers a particular traceable assessment of practical human control. These short descriptions emphasize differences in this comparison, not the full scope of each approach.

## 3. One source is not several independent witnesses

For the Tornado case, keep this reading map in mind:

**Official MoD inquiry summary → McCarthy's later analysis**

**Official inquiry material → TAE's later assessment**

McCarthy also draws on the US Defense Science Board report. The thesis's event sequence explicitly draws on the MoD summary. Two analyses citing the same summary do not turn that summary into two independent factual sources.

The thesis is valuable **related analysis**. It is not independent corroboration of every fact or of TAE's conclusions.

[McCarthy, Chapter 6 and bibliography](https://www.mit.stamp-institute.com/uploads/2013_A_System_Theoretic_Safety_Analysis_of_Fr.pdf) · [Official MoD summary](https://assets.publishing.service.gov.uk/media/5a78e39b40f0b62b22cbd9a5/maas03_02_tornado_zg710_22mar03.pdf)

## 4. A careful example: what does “unresolved” mean here?

The ZG710 event involved the deaths of two RAF crew members. It deserves care, not treatment as merely a convenient test case.

### A. What the official source says
The MoD summary identifies training and doctrine, and autonomous battery operation, among contributory factors. It describes rapid-engagement training and the limited wider air picture. See **paragraphs 11–12** of the [official summary](https://assets.publishing.service.gov.uk/media/5a78e39b40f0b62b22cbd9a5/maas03_02_tornado_zg710_22mar03.pdf).

### B. What our narrower assessment asks
Does the particular evidence support each condition required by TAE for the specific decision and time period being assessed? An adverse description of training does not, by itself, establish the particular failure judgment required by that assessment. The inference needs to be shown, not assumed.

### C. What we must not conclude
TAE's unresolved result **does not overturn the inquiry's contributory findings**. The two assessments ask different questions. Equally, we cannot use an inquiry's group-level account to claim knowledge of an individual's thoughts.

### D. A reasonable alternative reading
A reader may regard the inquiry's documented training and communications limitations as strong evidence of impaired practical control. The TAE reading may still withhold a binary result because its required link to the bounded condition is not sufficiently established. That disagreement belongs in the record, not hidden behind a label.

**This is an explanatory comparison, not a new scoring decision.** In the historical study book, open `02_ZG710_WORKED_EXAMPLE.ipynb` for the preserved original judgment, correction and source links. No historical record has been changed here.

## 5. So what can we honestly say this study contributes?

A cautious working description is:

> TAE presents an inspectable implementation of a bounded human-control assessment. It shows how selected public records lead to assessments of each requirement, how corrections change the result, and where the evidence remains unresolved. Its usefulness beyond these demonstrations still requires evaluation.

That is narrower than “we proved a new, superior way to measure human control.” It is also clearer and more defensible.

The November paper already acknowledges substantial overlap with related work. This companion makes that acknowledgment easier to follow. It does not supply a first-ever claim, a comparative benchmark or a validated benefit.

**A useful reader check:** Can you locate the source, distinguish the author's interpretation from it, and see why the conclusion stops where it does? A clear notebook helps with inspection; it does not itself prove the method is valid.
