## What question are we trying to answer?

Suppose an institution says that a person had control over an automated decision. What records would let another reader examine that claim?

The paper asks how to organize the documentary evidence for one bounded decision. It does not directly observe private thinking. A person's title, an approval, or the eventual outcome cannot answer every question about control.

The procedure connects a source record to an assessor's interpretation, then to a condition label and an overall result. Its value for a reader is that the path can be inspected and challenged. Whether other assessors agree or institutions benefit remains unestablished.

**The distinction to remember:** a reproducible calculation can follow an uncertain judgment. Code does not make that judgment true.

## Six ordinary questions before the technical labels

| Ask first | The paper calls this |
|:---|:---|
| What relevant information reached the person before commitment? | Information access |
| What evidence shows a capacity to interpret it under the actual conditions? | Comprehension capacity |
| What was the person permitted to change? | Intervention authority |
| Was there a usable opportunity and path to act? | Intervention feasibility |
| What judgment or intervention is actually recorded? | Exercised judgment |
| What connects that intervention to execution? | Execution propagation |

These questions are related, but they are not interchangeable. Receiving information is not proof of understanding. Permission is not proof that it was usable. A harmful outcome does not by itself prove that an issued action failed to reach execution.

This is the study's bounded assessment contract, not a validated universal definition of human control.

## One example: a bad outcome is not the same as a failed action link

In the preserved F/A-18C assessment, the execution-propagation field was **Unsupported**. Its exact recorded rationale was:

This quotation is **the assessor's wording**, not a quotation from an inquiry or a model response.

The rationale concerns whether harm was prevented. But the released propagation question asks whether an intervention reached execution or changed obligations. Those are different propositions. Correcting that mismatch does not establish that the intervention was sound or that control was complete.

The limited correction retained **P/I alternatives** for propagation: either some positive component is supported with a material gap, or the record is insufficient to decide. ## What do the labels mean?

- **Supported (S):** the packet meets the released direct-contemporaneous evidence requirement for the condition. This is an assessment under the rule, not a guarantee of historical truth.
- **Partially supported (P):** some required elements are present, but a material gap remains.
- **Unsupported (U):** evidence contradicts the condition or shows it was absent.
- **Indeterminate (I):** the packet lacks enough evidence to decide.
- **P/I:** two retained alternatives awaiting a justified choice. It is not a probability or a new category.

For the six required conditions, the rule is: **any U → Fail; all S → Pass; otherwise P or I → Unresolved**. Missing information elsewhere cannot cancel an established U.

“Fail” and “Unresolved” describe the bounded evidence assessment. Neither is a verdict about a person's character or culpability.

## Case 1: Oko asks what a later account can establish

The preserved packet concerns Stanislav Petrov's response to an Oko warning in 1983. The assessment relies on later accounts rather than a complete contemporaneous operator record.

**Exact recorded access rationale:**

> Retrospective accounts report that Petrov received the alert and checked contextual and ground-radar information. The packet contains no contemporaneous delivery, interface, or command record.

The v0.17 comparison uses the revised **v0.6 Oko record**, where all six conditions are P. Do not substitute the earlier v0.3 record, which had six S. The separate Oko correction predates the three-way comparison shown below.

**The reading question:** which part is reported by a source, and which part requires a contemporaneous record that the packet does not contain? An account can be informative without meeting every requirement of this protocol.

## Case 2: ZG710 asks how far an adverse finding supports an inference

The UK Ministry of Defence summary reports a Patriot battery crew launch after tracking and IFF interrogation (§4, printed page 2). It also describes training, time and communications constraints (§§11–12, printed page 3). Those source statements matter. They do not directly record the Patriot battery crew's private thinking.

The historical assessment treated comprehension as Unsupported. The approved later reassessment treated it as Indeterminate because the inference to a necessary capacity failure was insufficiently established. It did not remove the adverse evidence or establish that comprehension was present.

**The reading question:** does a documented preparation deficiency establish that the Patriot battery crew lacked the necessary capacity, or does that inference need further support? The worked example preserves the strongest alternative interpretation instead of declaring it defeated by missing evidence.

Two RAF aircrew died. The review concerns the limits of our assessment, not a reassessment of their worth or an accusation against the people involved.

## Case 3: F/A-18C asks what an incomplete record leaves open

The packet concerns a Patriot engagement of a U.S. Navy F/A-18C in 2003. The preserved assessment describes gaps in the record needed to examine intervention feasibility.

**Exact recorded feasibility rationale:**

> Missing timing, display, communication, and operator records prevent a finding about whether an effective challenge was feasible.

The case's limited correction changed the harm-based propagation inference and retained uncertainty about authority. Several other conditions were already Indeterminate. The resulting assessment is Unresolved under the retained inputs.

**The reading question:** what record would distinguish an absent opportunity from an opportunity that simply is not documented in the public packet? We should not assume such records exist or can be obtained.

The fatal outcome must not be used as a substitute for this evidence. This guide has not newly verified every F/A-18C source. ## Why do the results change?

These rows are different assessment checkpoints, **not new events or independent datasets**.

| Preserved comparison | Oko | ZG710 | F/A-18C |
|:---|:---|:---|:---|
| Historical v0.17 baseline | Unresolved | Fail | Fail |
| Limited correction | Unresolved | Fail | Unresolved |
| Approved additional reassessment | Unresolved | Unresolved | Unresolved |

Correcting propagation alone did not remove ZG710's other Unsupported conditions. The additional reassessment separately addressed comprehension, feasibility and exercise. Its outcome is conditional on the approved judgments and retained alternatives.

The reported 64 assignments enumerate those retained alternatives. They are not 64 independent cases, and they do not test every defensible interpretation. If a necessary U is justified, the rule still requires Fail.

The latest row does not show that control was present, absent, safe or unsafe. It shows what this record supports under the stated assessment choices. [Approved input record](sources/additional_inputs.json) · [approval scope and limits](sources/public_author_approval.json).

## What did this study accomplish?

It produced a traceable demonstration: a reader can locate the evidence used, inspect the condition judgments, see the derived results and follow corrections without erasing earlier assessments. That is something concrete to examine and improve.

It did not establish classification accuracy, independent-assessor agreement, population prevalence, institutional benefit, or transfer to contemporary AI agents. The three cases were purposefully selected under predefined strata, not sampled to represent all automated decisions. The two Patriot cases should not be treated as independent evidence of a general rate.

Existing research already addresses oversight, uncertainty, reconstruction and assurance. This study's implemented combination is open to comparison; it should not be presented as the first discovery of those problems.

## Choose your next step

**If you are learning the method:** read the [ZG710 worked example](02_ZG710_WORKED_EXAMPLE.ipynb). It follows one disputed condition in detail and keeps source statements, interpretations and competing arguments separate.

**If you want the full preserved case record:** open [Complete case records](01_COMPLETE_CASE_RECORDS.ipynb). It includes all 18 baseline condition rationales, source manifests and the later correction material. Old provisional statements remain labeled as historical.

**If you want to assess a claim:** ask what precise source supports it, what the assessor inferred, what alternative remains, and what evidence would distinguish the alternatives. You do not need to agree with a label to understand its derivation.

**If you are thinking about the next paper:** first identify one point this reader makes clearer and one inference you still want to inspect. The purpose now is understanding. Shortening the paper is not a prerequisite or a new scientific result.
