## Why the limited correction mattered

The following is an exact excerpt from the preserved derivation ledger inside the historical ZIP, from **Correction 2A** through **Correction 2B**. Its provisional author-review language describes that earlier checkpoint. The later public approval and November manuscript retain their separate scope. The excerpt is assessor reasoning, not new primary evidence.

> Correction 2A Harm is not evidence that a decision failed to propagate
> 
> Changed assessments: T /practical_control/effect, U → {P or I}; F /practical_control/effect, U → {P or I}, pending adjudication of the positive component under the released definition.
> 
> Exact old targets and source premises are in the two Patriot propagation rows below. Both old rationales use failure to prevent irreversible engagement. The pinned v0.16 protocol already excludes beneficial-outcome interpretation of effect, and the earlier effect question concerned alteration of execution/obligations. T2 §4 supplies an affirmative crew-to-launch account; F3's identified sequence supplies a reported ordering/engagement account. Neither the fatal outcome nor those old rationales establish that a relevant instruction was ignored or misimplemented. No inspected passage supplies a contrary-command implementation failure sufficient to rescue those U rationales.
> 
> What is established here is the mismatch between the stated negative rationale and the released propagation condition. What is not uniquely settled is P versus I: the author must decide whether the available record establishes a genuine required positive component at the original granularity. A launched missile alone is not such a component. Do not infer S from a later reconstruction while retaining the strict original S threshold. Do not change the claim to effective prevention and call that an unchanged propagation assessment.
> 
> Change type: E. No new construct, scope or aggregation is required to reject harm alone as a propagation contradiction. Choosing a narrower action or accepting a later source as complete proof belongs in Stage 3, not here.
> 
> Correction 2B Apply the released S evidence threshold consistently
> 
> Changed assessments: T /practical_control/authority, S → {P or I}; F /practical_control/authority, S → {P or I}, pending the released P/I positive-element judgment.
> 
> The baseline S rule requires direct contemporaneous evidence. T cites T2/T3/T4; F cites F3/F4. The inspected materials are later findings/reporting, and the actual event-period delegation or rights records are not reproduced as inspected original evidence in this packet. An inquiry's access to such evidence is not the assessor's inspection of it. The Oko v0.6 temporal correction cannot be applied only to participant accounts while later official findings silently satisfy the same temporal gate. T2 §4/§12 supports a reconstructed firing role/right; the F3 ordering sequence supports a reported command role. Whether that amounts to a material positive element for P under the incompletely specified old decomposition remains an author judgment; otherwise use I.
